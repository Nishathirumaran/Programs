```python
a = 10
b = 3
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)
```

Arithmetic operators

```python
x = 40
x += 2
print(x)
x -= 2
print(x)
x *= 2
print(x)
x /= 2
print(x)
x %= 3
print(x)
x **= 2
print(x)
x //= 6
print(x)
```

Assignment operators

```python
a = 2
b = 30
print(a==b)
print(a!=b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
```

Comparison operators

```python
a = 30
print(a>10 and a<10)
print(a==2 or a!=3)
print(not(a<2))
```

Conditional operators

```python
a = [1,2,3]
b = [4,2,1]
c = b
print(a is b)
print(a is not b)
print(a is c)
print(a is not c)
print(b is c)
print(b is not c)
```

Identity operators

```python
vehicle = ["car", "bike", "motorcycle"]
print("bus" in vehicle)
print("car" not in vehicle)
```

Membership operators

```python
a = 2
b = 30
print(a&b)
print(a/b)
print(a^b)
```
