```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[0])
print(arr[-1])
```

```python
arr = np.array([10, 20, 30, 40, 50, 65, 87])
print(arr[arr % 2 == 0])
```

```python
arr = np.array([10, 20, 30, 40, 50, 25, 35, 15])
print(arr[arr % 2 != 0])
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.sum(arr))
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.mean(arr))
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.max(arr))
print(np.min(arr))
```

```python
arr = np.array([10, 20, 30, 40, 50, 34, 15, 38])
print(np.sort(arr))
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[::-1])
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr * 2)
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr + 10)
```

```python
arr = np.array([10, 20, 30, 40, 50, 65, 79, 87, 90])
print(arr[arr > 50])
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.where(arr == 30))
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr.size)
```
